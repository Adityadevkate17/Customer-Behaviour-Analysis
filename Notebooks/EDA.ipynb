## `gender`- `subscription_status`-`purchase_amount`

```python
# Total Purchase Amount
df.groupby(['gender', 'subscription_status'])['purchase_amount'].sum().unstack()
```

```python
# Customer Count
df.groupby(['gender', 'subscription_status']).size().unstack()
```

### Obserrvation 
- Male customers spend more overall than female customers. Male customers with subscriptions contribute 62,645 in purchases, while female customers have no subscription records in this output.

## `category` - `subscription_status` `purchase_amount`

```python
df.groupby(['category', 'subscription_status'])['purchase_amount'].mean().unstack().plot(kind='bar')
plt.title('Category vs Subscription Status vs Average Purchase Amount')
plt.ylabel('Average Purchase Amount')
plt.show()
```

### Observations 
- Clothing has the highest total purchase amount among all categories
- the average purchase amount is similar for both subscribers and non-subscribers across all categories.

## `discount_applied` - `gender` - `purchase_amount`

```python
df.groupby(['discount_applied', 'gender'])['purchase_amount'].mean().unstack()
```

```python
df.groupby(['discount_applied', 'gender'])['purchase_amount'].sum().unstack()
```

```python
df.groupby(['discount_applied', 'gender'])['purchase_amount'].mean().unstack().plot(kind='bar')
plt.title('Discount vs Gender vs Average Purchase Amount')
plt.ylabel('Average Purchase Amount')
plt.show()
```

### Observation 
- The average purchase amount is almost similar for subscribers and non-subscribers across all categories.
